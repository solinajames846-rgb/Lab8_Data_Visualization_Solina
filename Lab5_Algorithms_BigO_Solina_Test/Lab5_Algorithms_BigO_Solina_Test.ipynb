```python
import time
import random

# Student Parameters
LAST_NAME = "Solina"
STUDENT_ID = "TUPM-26-2302"
FAVORITE_ARTIST = "BINI"

# Derived Values per Laboratory Rules
SEED_DIGIT = int(STUDENT_ID[-1])                       # 2
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # 15
NAME_LENGTH = len(LAST_NAME)                           # 6
ARTIST_LENGTH = len(FAVORITE_ARTIST)                   # 4
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST) # 2+9+14+9 = 34

print("="*60)
print(f"LABORATORY ACTIVITY 5 - ALGORITHMS & BIG O")
print(f"Student: {LAST_NAME.upper()} | ID: {STUDENT_ID}")
print(f"Values: SEED_DIGIT={SEED_DIGIT}, ID_SUM={ID_SUM}, ARTIST_SUM={ARTIST_SUM}")
print("="*60)
```

```python
print("--- EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH ---")

# 1. Generate unique records sorted array
original_dataset = [10 + i * (SEED_DIGIT + 2) for i in range(15)]
rotation_point = SEED_DIGIT if SEED_DIGIT > 0 else 3

# Perform rotation
rotated_dataset = original_dataset[rotation_point:] + original_dataset[:rotation_point]
target_incident = ID_SUM + ARTIST_SUM # 15 + 34 = 49

# Linear Search Implementation
def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

# Modified Binary Search for Rotated Array
def rotated_binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, comparisons
        
        # Left half is normally sorted
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Right half is normally sorted
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

# Execute Searches
pos_linear, comps_linear = linear_search(rotated_dataset, target_incident)
pos_binary, comps_binary = rotated_binary_search(rotated_dataset, target_incident)

print(f"\n[Assessment Data]")
print(f"Original Dataset : {original_dataset}")
print(f"Rotated Dataset  : {rotated_dataset} (Rotated at index {rotation_point})")
print(f"Target Incident  : {target_incident}")
print(f"Comparison Counts: Linear Search = {comps_linear} | Rotated Binary Search = {comps_binary}")

print(f"\n[Final Output]")
if pos_binary != -1:
    print(f"Target found at index position: {pos_binary}")
else:
    print("Target incident was not found in the records.")

print("\n[Complexity Analysis]")
print("Linear Search        -> Time: O(N), Space: O(1)")
print("Rotated Binary Search-> Time: O(log N), Space: O(1)")
```

```python
print("--- EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS ---")

# 1. Dataset Generation
dataset_A = sorted([100 + i * SEED_DIGIT for i in range(8)]) 
dataset_B = sorted([105 + i * ARTIST_LENGTH for i in range(6)])

# Two-Pointer Consolidation (Merge Function)
def merge_sorted_collections(arr1, arr2):
    merged = []
    i = j = comparisons = 0
    
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
            
    # Append remaining
    merged.extend(arr1[i:])
    merged.extend(arr2[j:])
    return merged, comparisons

consolidated_dataset, total_comps = merge_sorted_collections(dataset_A, dataset_B)
is_sorted = consolidated_dataset == sorted(consolidated_dataset)

print(f"\n[Assessment Data]")
print(f"Department A dataset : {dataset_A}")
print(f"Department B dataset : {dataset_B}")
print(f"Consolidated dataset : {consolidated_dataset}")
print(f"Number of records    : Before={len(dataset_A)+len(dataset_B)} | After={len(consolidated_dataset)}")
print(f"Comparison count     : {total_comps}")
print(f"Ordering verification: {'PASSED (Sorted)' if is_sorted else 'FAILED'}")

print(f"\n[Final Output]")
print(f"Consolidated Enrollment Queue: {consolidated_dataset}")
print("\n[Complexity Analysis]")
print("Two-pointer Merge Strategy -> Time: O(N + M) | Space: O(N + M)")
```

```python
print("--- EXERCISE 3: E-COMMERCE ORDER PROCESSING ---")

# Generate base array using student numbers
base_orders = [int((ARTIST_SUM * i + ID_SUM) % 200) for i in range(1, 31)]

# Sorting Algorithm Variants with metrics track
def bubble_sort(arr_input):
    arr = arr_input.copy()
    comps = moves = 0
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            comps += 1
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                moves += 1
    return comps, moves

def selection_sort(arr_input):
    arr = arr_input.copy()
    comps = moves = 0
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            comps += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            moves += 1
    return comps, moves

def insertion_sort(arr_input):
    arr = arr_input.copy()
    comps = moves = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1
        while j >= 0:
            comps += 1
            if arr[j] > key:
                arr[j+1] = arr[j]
                moves += 1
                j -= 1
            else:
                break
        arr[j+1] = key
    return comps, moves

# Prepare Variations
inputs = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print(f"\n[Assessment Data - Performance Matrix]")
print(f"{'Algorithm':<15} | {'Condition':<16} | {'Comparisons':<12} | {'Movements':<10}")
print("-" * 65)

for alg_name, alg_func in algorithms.items():
    for cond_name, test_arr in inputs.items():
        start_time = time.perf_counter()
        comps, moves = alg_func(test_arr)
        duration = (time.perf_counter() - start_time) * 1000
        print(f"{alg_name:<15} | {cond_name:<16} | {comps:<12} | {moves:<10}")
```
